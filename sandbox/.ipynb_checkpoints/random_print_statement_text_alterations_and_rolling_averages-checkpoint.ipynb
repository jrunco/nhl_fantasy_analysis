```python
from IPython.display import display, HTML

''' basic changes '''
### bold, underlined, etc.
print('This text is normal')
print('\033[1m' + 'This text will be bold!' + '\033[0m')
print("\033[4mThis text will be underlined\033[0m")
print("\033[1;4m" + "This is bold and underlined text" + "\033[0m")

### text color
print("\033[31m" + "This text is Red" + "\033[0m")
print("\033[32m" + "This text is Green" + "\033[0m")
print("\033[34m" + "This text is Blue" + "\033[0m")
print("\033[33m" + "This text is Yellow" + "\033[0m")


### multiple changes at once
text = "Your printed text here"
display(HTML(f'<span style="font-size: 12px;">{text}</span>'))
display(HTML(f'<span style="font-size: 20px;">{text}</span>'))
display(HTML(f'<span style="font-size: 30px;">{text}</span>'))
display(HTML(f'<span style="font-size: 30px;">Your printed text here</span>'))
display(HTML(f"<p style='font-size:30px; font-weight:bold;'>{text}</p>"))
display(HTML('<span style="font-size: 30px; font-weight: bold; text-decoration: underline;">Your printed text here</span>'))
display(HTML('<span style="font-size: 30px; color: red; font-weight: bold; text-decoration: underline;">Your printed text here</span>'))

```

# Option 2 should work for versions of pandas newer than v1.5.0

```python
import pandas as pd

# Create sample data with a datetime index
dates = pd.date_range(start="2026-01-01", end="2026-03-01", freq="D")
df = pd.DataFrame({"value": range(len(dates))}, index=dates)

# Option 1: Rolling mean with a 14-day window, sampled every 7 days
weekly_overlapping_avg = df.rolling(window="14d").mean().resample("7d").last()

# Option 2: Using the step parameter (Pandas 2.0+)
step_avg = df.rolling(window="14d", step="7d").mean()

"""
Key Parameterswindow='14d': Sets the window size to cover a two-week span.resample('7d') or 
step='7d': Moves the window forward by one week at a time, creating the 50% (one-week) overlap.
"""
```

```python
import numpy as np
import pandas as pd

# 1. Create a dummy time-series DataFrame
dates = pd.date_range(start="2026-01-01", end="2026-03-01", freq="D")
df = pd.DataFrame({"value": np.random.randint(10, 100, size=len(dates))}, index=dates)

# 2. Resample to the overlap frequency (1 week) using the weekly mean
# 'W' handles the 1-week step/overlap interval
weekly_resampled = df.resample("W").mean()

# 3. Apply a rolling window of 2 periods (2 weeks total)
# window=2 over weekly resampled data sums up to a 14-day window
two_week_rolling_avg = weekly_resampled.rolling(window=2).mean()

print(two_week_rolling_avg)
```

```python
import pandas as pd

# 1. Sample data with missing days
data = {
    "date": pd.to_datetime(
        [
            "2026-01-01",
            "2026-01-03",
            "2026-01-10",
            "2026-01-15",
            "2026-01-20",
            "2026-01-25",
        ]
    ),
    "value": [10, 20, 30, 40, 50, 60],
}

df = pd.DataFrame(data)

# 2. Set index as datetime and sort
df = df.set_index("date").sort_index()

# 3. Calculate 14-day window with 7-day step (1-week overlap)
result = df.rolling(window="14D", step="7D").mean()

print(result)
```
